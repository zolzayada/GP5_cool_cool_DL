**Очистка данных**

Перейдем к очистке данных. Для начала удалим столбец police_id, так как это просто техничсеких айдишник, не несущий никакой пользы для обучения модели.

```python
df = df.drop('policy_id', axis=1)
```

При первичном анализе датасета мы заметили, что признаки max_power (мощность двигателя) и max_torque (крутящийся момент) имеют тип object и хранятся как строки с дополнительными символами, хотя по сути являются числовыми характеристиками двигателя. Проверим так ли это и если да, то преобразуем в числовые признаки.

```python
df['max_power'].head(10)
```

```python
# Нам нужно взять только то, что перед bhp
df['max_power'] = (df['max_power'].str.split('bhp').str[0].astype(float))
```

```python
df['max_power'].head(10)
```

```python
df['max_torque'].head(10)
```

```python
# Нам нужно взять только то, что перед Nm
df['max_torque'] = (df['max_torque'].str.split('Nm').str[0].astype(float))
```

```python
df['max_torque'].head(10)
```

Дальше нужно посмотреть, есть ли у нас выбросы. Посмотрим на числовые колонки, чтобы их найти.

```python
numbers = df.select_dtypes(include=['number'])
numbers.columns
```

Отсюда можем исключить: is_claim - таргет, make - закодированный производитель (не имеет количественного смысла), airbags, cylinder, gear_box - дискретные признаки (редкие значения тут могут являться корректными характеристиками автомобиля, их не надо считать за выбросы), ncap_rating - рейтинг (аналогично дискретным).

```python
numbers = numbers.drop(['is_claim', 'make', 'airbags', 'cylinder', 'gear_box', 'ncap_rating'], axis = 1)
```

```python
numbers.describe()
```

В целом, по первичному осмотру особо никаких подозрительных значений в глаза не бросается. Везде не особо большой разброс между минимум и нижним квартилем И максимумом и верхним квартилем. Большой только в age_of_car и age_of_policyholder, но это потому, что это нормализованные уже значения и поэтому максимум там 1, что логично.

```python
plt.figure(figsize=(20,6))
sns.boxplot(data=numbers)
plt.show()
```

По BoxPlot видим, что предположительно выбросы есть только у population_density, но в целом такие значение (60000-70000) норм для мегаполисов. А так как больших мегаполисов мало, вероятно, и таких значений мало (как у нас Москва, например, очень сильно отличается по плотности населения от других городов Росии). Так что, думаю, стоит оставить эти значения как есть. По age_of_car и age_of_policyholder, как и прописано выше, не рассматриваем, так как это нормализованные значения уже.

**Категориальные данные**

у нас много бинарных признаков со значениями Yes и No - закодируем их.

```python
columns = ['is_esc', 'is_adjustable_steering', 'is_tpms', 'is_parking_sensors', 'is_parking_camera', 
           'is_front_fog_lights', 'is_rear_window_wiper', 'is_rear_window_washer', 'is_rear_window_defogger', 
           'is_brake_assist', 'is_power_door_locks', 'is_central_locking', 'is_power_steering', 'is_driver_seat_height_adjustable', 
           'is_day_night_rear_view_mirror', 'is_ecw', 'is_speed_alert']
for col in columns:
    df[col] = df[col].map({'Yes':1,'No':0})

df.head()
```

Для полносвязаной нейронной сети нам необходимо закодировать все остальные категории тоже.

```python
categor = df.select_dtypes(include='object')
categor.columns
```

```python
categor.nunique()
```

У нас не супер много категорий различных, поэтому можем использовать one hot encoding https://pandas.pydata.org/docs/reference/api/pandas.get_dummies.html.

```python
df = pd.get_dummies(df, columns=categor.columns, drop_first=True, dtype=int)

df.head()
```
